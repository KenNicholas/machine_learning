```python
train=train.dropna(axis=1)
train.dtypes
```

```python
Input_Attribute = [f for f in train.columns if train.dtypes[f] != 'object']
Input_Attribute.remove('Id')
Input_Attribute.remove('SalePrice')
```

```python
len(Input_Attribute)
```

```python
Input_Attribute
```

```python
x = train[Input_Attribute]
x.describe()
```

```python
y=train.SalePrice
y.shape
```

```python
x.shape
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x,y)
```

```python
model.score(x,y)
```

```python
y_pred = model.predict(x)
```

```python
from sklearn.metrics import r2_score
r2_score(y, y_pred)
```

```python
from sklearn.metrics import mean_squared_error
mean_squared_error(y, y_pred)
```

```python
from sklearn.metrics import mean_absolute_error
mean_absolute_error(y, y_pred)
```

```python
from sklearn.tree import DecisionTreeRegressor
modelDTR = DecisionTreeRegressor()
modelDTR.fit(x,y)
modelDTR.score(x,y)
```
